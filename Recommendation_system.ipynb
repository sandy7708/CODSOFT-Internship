```python
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity

data = {
    'User': ['A', 'A', 'A', 'B', 'B', 'C', 'C', 'D', 'D'],
    'Movie': ['Inception', 'Titanic', 'Avatar',
              'Inception', 'Avatar',
              'Titanic', 'Avatar',
              'Inception', 'Titanic'],
    'Rating': [5, 3, 4, 5, 5, 4, 5, 4, 5]
}

df = pd.DataFrame(data)

user_movie_matrix = df.pivot_table(
    index='User',
    columns='Movie',
    values='Rating'
).fillna(0)

similarity = cosine_similarity(user_movie_matrix)

similarity_df = pd.DataFrame(
    similarity,
    index=user_movie_matrix.index,
    columns=user_movie_matrix.index
)

def recommend_movies(user_name, top_n=2):
    similar_users = similarity_df[user_name].sort_values(ascending=False)
    similar_users = similar_users.drop(user_name)

    watched = user_movie_matrix.loc[user_name]
    watched_movies = watched[watched > 0].index.tolist()

    recommendations = {}

    for sim_user in similar_users.index:
        sim_user_movies = user_movie_matrix.loc[sim_user]

        for movie in sim_user_movies.index:
            if sim_user_movies[movie] > 0 and movie not in watched_movies:
                recommendations[movie] = recommendations.get(movie, 0) + sim_user_movies[movie]

    recommended_movies = sorted(
        recommendations.items(),
        key=lambda x: x[1],
        reverse=True
    )

    return [movie for movie, score in recommended_movies[:top_n]]

user = 'B'

print("Recommended movies for User", user)

for movie in recommend_movies(user):
    print(movie)
```
